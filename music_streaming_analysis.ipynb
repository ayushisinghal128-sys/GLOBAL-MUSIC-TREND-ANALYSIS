# Global Music Streaming Trends Analysis
End-to-end EDA using Python.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/Global_Music_Streaming_Listener_Preferences.csv')
df.head()
```

```python
df.shape
df.info()
df.isna().sum()
df.duplicated().sum()
```

```python
df.describe(include='all').T
```

```python
df['Streaming Platform'].value_counts().plot(kind='bar', title='Users by Streaming Platform')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Top Genre'].value_counts().plot(kind='bar', title='Users by Top Genre')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
order = ['Morning','Afternoon','Night']
df['Listening Time (Morning/Afternoon/Night)'].value_counts().reindex(order).plot(kind='bar', title='Preferred Listening Time')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
country_avg = df.groupby('Country')['Minutes Streamed Per Day'].mean().sort_values(ascending=False)
country_avg.plot(kind='bar', title='Average Minutes Streamed Per Day by Country')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
age_groups = pd.cut(df['Age'], bins=[0,20,30,40,50,60,100], labels=['<=20','21-30','31-40','41-50','51-60','61+'])
age_groups.value_counts().sort_index()
```

```python
platform_engagement = df.groupby('Streaming Platform')[['Discover Weekly Engagement (%)','Repeat Song Rate (%)']].mean().sort_values('Repeat Song Rate (%)', ascending=False)
platform_engagement
```
